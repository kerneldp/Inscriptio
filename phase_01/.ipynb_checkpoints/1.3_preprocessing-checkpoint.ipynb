# 1.3 — Core Image Standardization Pipeline
Implements `ImagePreprocessor`: grayscale -> Otsu binarisation ->
foreground-background inversion -> bilinear resize 224x224 -> float32 [0,1].
Batch processes data/raw/mendeley/ -> data/processed/.
Validates shape (224,224,1), dtype float32, pixel range [0,1].

```python
import cv2, numpy as np, pathlib

RAW_DIR       = pathlib.Path('../../data/raw/mendeley/DATASET DYSGRAPHIA HANDWRITING')
PROCESSED_DIR = pathlib.Path('../../data/processed')
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
TARGET_SIZE = (224, 224)
print(f'Raw dir    : {RAW_DIR}')
print(f'Output dir : {PROCESSED_DIR}')
```

```python
class ImagePreprocessor:
    '''Standardises raw handwriting images to 224x224 float32 tensors.'''

    def __init__(self, target_size=(224, 224)):
        self.target_size = target_size

    def to_grayscale(self, bgr_img):
        '''Convert BGR image to single-channel grayscale.'''
        return cv2.cvtColor(bgr_img, cv2.COLOR_BGR2GRAY)

    def binarise(self, gray_img):
        '''Apply Otsu threshold: cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU).'''
        _, binary = cv2.threshold(gray_img, 0, 255,
                                  cv2.THRESH_BINARY + cv2.THRESH_OTSU)
        return binary

    def invert(self, binary_img):
        '''Invert: white strokes on black bg via THRESH_BINARY_INV + THRESH_OTSU.'''
        _, inverted = cv2.threshold(binary_img, 0, 255,
                                    cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
        return inverted

    def resize(self, img):
        '''Bilinear resize: cv2.resize(inverted, (224,224), interpolation=cv2.INTER_LINEAR).'''
        return cv2.resize(img, self.target_size, interpolation=cv2.INTER_LINEAR)

    def normalise(self, img):
        '''Divide by 255.0 -> float32 tensor in [0.0, 1.0].'''
        return (img / 255.0).astype(np.float32)

    def __call__(self, image_path):
        '''Full pipeline. Returns ndarray shape (224,224,1), dtype float32.'''
        bgr = cv2.imread(str(image_path))
        if bgr is None:
            raise FileNotFoundError(f'Cannot read: {image_path}')
        gray     = self.to_grayscale(bgr)
        binary   = self.binarise(gray)
        inverted = self.invert(binary)
        resized  = self.resize(inverted)
        normed   = self.normalise(resized)
        return np.expand_dims(normed, axis=-1)  # (224,224,1)

preprocessor = ImagePreprocessor(target_size=TARGET_SIZE)
print('ImagePreprocessor instantiated.')
```

```python
# Batch processing: data/raw/mendeley/ -> data/processed/
CLASS_DIRS = [('Low Potential Dysgraphia', 0), ('Potential Dysgraphia', 1)]
errors = []
processed_count = 0

for class_name, label in CLASS_DIRS:
    src_dir = RAW_DIR / class_name
    dst_dir = PROCESSED_DIR / class_name
    dst_dir.mkdir(parents=True, exist_ok=True)
    images = sorted(src_dir.glob('*.jpg')) + sorted(src_dir.glob('*.jpeg')) + sorted(src_dir.glob('*.png'))
    print(f'Processing {class_name}: {len(images)} images ...')
    for img_path in images:
        try:
            tensor = preprocessor(img_path)
            out_path = dst_dir / (img_path.stem + '.npy')
            np.save(str(out_path), tensor)
            processed_count += 1
        except Exception as e:
            errors.append((str(img_path), str(e)))

print(f'Processed: {processed_count} images')
if errors:
    print(f'ERRORS: {len(errors)}')
    for path, err in errors:
        print(f'  {path}: {err}')
else:
    print('No errors. All images processed successfully.')
```

```python
# Validate all output tensors: shape (224,224,1), dtype float32, pixel range [0,1]
all_npy = list(PROCESSED_DIR.rglob('*.npy'))
print(f'Validating {len(all_npy)} output tensors ...')
validation_errors = []
for npy_path in all_npy:
    arr = np.load(str(npy_path))
    if arr.shape != (224, 224, 1):
        validation_errors.append(f'Shape: {npy_path} -> {arr.shape}')
    if arr.dtype != np.float32:
        validation_errors.append(f'Dtype: {npy_path} -> {arr.dtype}')
    if arr.min() < 0.0 or arr.max() > 1.0:
        validation_errors.append(f'Range: {npy_path} -> [{arr.min():.4f},{arr.max():.4f}]')

if validation_errors:
    for e in validation_errors: print(f'  FAIL: {e}')
    raise AssertionError(f'{len(validation_errors)} validation errors!')
else:
    print(f'All {len(all_npy)} tensors: shape (224, 224, 1), dtype float32, pixel range [0.0, 1.0]')
```
