# 1.2 — Dataset Partitioning
Splits Mendeley (LPD+PD) into Train/Val/Test at 50/25/25.
- Test set: original, unaugmented Mendeley images ONLY.
- Val set: preprocessed but non-augmented images.
- Augmentation applied to Training split only (see 1.4).
Writes manifests and logs class imbalance to reports/dataset.md.

```python
import os, csv, random, pathlib
from collections import Counter, defaultdict

SEED = 42
random.seed(SEED)

BASE_DIR     = pathlib.Path('../../data')
RAW_DIR      = BASE_DIR / 'raw' / 'mendeley' / 'DATASET DYSGRAPHIA HANDWRITING'
MANIFEST_DIR = BASE_DIR / 'manifests'
REPORTS_DIR  = pathlib.Path('../../reports')
MANIFEST_DIR.mkdir(parents=True, exist_ok=True)
REPORTS_DIR.mkdir(parents=True, exist_ok=True)

CLASS_DIRS = {
    'Low Potential Dysgraphia': 0,
    'Potential Dysgraphia':     1,
}
TRAIN_RATIO, VAL_RATIO, TEST_RATIO = 0.50, 0.25, 0.25
print(f'RAW_DIR exists: {RAW_DIR.exists()}')
print(f'Split: train={TRAIN_RATIO} / val={VAL_RATIO} / test={TEST_RATIO}')
```

```python
all_samples = []
for class_name, label in CLASS_DIRS.items():
    class_dir = RAW_DIR / class_name
    images = sorted([f for f in class_dir.iterdir()
                     if f.suffix.lower() in ('.jpg', '.jpeg', '.png')])
    print(f'  {class_name}: {len(images)} images (label={label})')
    for img in images:
        all_samples.append((str(img.resolve()), label))
print(f'Total: {len(all_samples)}')
```

```python
def stratified_split(samples, train_r, val_r, seed=42):
    by_class = defaultdict(list)
    for path, label in samples:
        by_class[label].append((path, label))
    train, val, test = [], [], []
    rng = random.Random(seed)
    for label, items in sorted(by_class.items()):
        rng.shuffle(items)
        n = len(items)
        n_train = int(n * train_r)
        n_val   = int(n * val_r)
        train.extend(items[:n_train])
        val.extend(items[n_train:n_train + n_val])
        test.extend(items[n_train + n_val:])
    return train, val, test

train_set, val_set, test_set = stratified_split(all_samples, TRAIN_RATIO, VAL_RATIO, seed=SEED)
print(f'Train: {len(train_set)}  Val: {len(val_set)}  Test: {len(test_set)}')
```

```python
def write_manifest(rows, filepath, augmented_flag=False):
    with open(filepath, 'w', newline='') as f:
        writer = csv.writer(f)
        writer.writerow(['image_path', 'label', 'augmented'])
        for path, label in rows:
            writer.writerow([path, label, augmented_flag])
    print(f'Written: {filepath}  ({len(rows)} rows)')

write_manifest(train_set, MANIFEST_DIR / 'train.csv', augmented_flag=False)
write_manifest(val_set,   MANIFEST_DIR / 'val.csv',   augmented_flag=False)
write_manifest(test_set,  MANIFEST_DIR / 'test.csv',  augmented_flag=False)
```

```python
# Assert zero data leakage
train_files = {r[0] for r in train_set}
val_files   = {r[0] for r in val_set}
test_files  = {r[0] for r in test_set}
assert len(train_files & val_files)  == 0, 'LEAKAGE train/val'
assert len(train_files & test_files) == 0, 'LEAKAGE train/test'
assert len(val_files   & test_files) == 0, 'LEAKAGE val/test'
print('ZERO data leakage confirmed across all three manifests.')
```

```python
# Log sample counts and class imbalance ratio
def class_counts(split):
    c = Counter(label for _, label in split)
    return c[0], c[1]

tr_lpd, tr_pd = class_counts(train_set)
va_lpd, va_pd = class_counts(val_set)
te_lpd, te_pd = class_counts(test_set)
tot_lpd = tr_lpd + va_lpd + te_lpd
tot_pd  = tr_pd  + va_pd  + te_pd
imbalance = max(tot_lpd, tot_pd) / min(tot_lpd, tot_pd)

report = f'''# Dataset Split Report

| Split | LPD (0) | PD (1) | Total |
|-------|---------|--------|-------|
| Train | {tr_lpd} | {tr_pd} | {len(train_set)} |
| Val   | {va_lpd} | {va_pd} | {len(val_set)} |
| Test  | {te_lpd} | {te_pd} | {len(test_set)} |
| **Total** | **{tot_lpd}** | **{tot_pd}** | **{len(all_samples)}** |

**Class Imbalance Ratio (majority/minority):** {imbalance:.4f}

## Notes
- Test: original unaugmented Mendeley images ONLY.
- Val: preprocessed but non-augmented Mendeley images.
- Augmentation applied to Training split only.
- Seed: {SEED}
'''

report_path = REPORTS_DIR / 'dataset.md'
with open(report_path, 'w') as f:
    f.write(report)
print(report)
print(f'Saved: {report_path}')
```
